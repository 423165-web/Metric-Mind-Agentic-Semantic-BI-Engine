```python
from faker import Faker
import pandas as pd
import random

fake = Faker()

# Generating the fake data for the customers table
customers = []

for i in range(1,1001):
    customers.append({
        "customer_id": i,
        "customer_name": fake.name(),
        "city": fake.city(),
        "country": fake.country(),
        "segment": random.choice(["Consumer","Corporate","Home Office"]),
        "created_date": fake.date_between("-5y","today")
    })

pd.DataFrame(customers).to_csv("customers.csv",index=False)
```

Generation of data for employee


```python
from datetime import date

fake = Faker()
Faker.seed(42)
random.seed(42)

departments = [
    "Sales",
    "Finance",
    "Human Resources",
    "Marketing",
    "IT",
    "Operations",
    "Customer Support"
]

titles = {
    "Sales": ["Sales Representative", "Sales Executive", "Sales Manager"],
    "Finance": ["Financial Analyst", "Accountant", "Finance Manager"],
    "Human Resources": ["HR Executive", "HR Manager", "Recruiter"],
    "Marketing": ["Marketing Executive", "Marketing Manager", "SEO Specialist"],
    "IT": ["Software Engineer", "Data Engineer", "Data Analyst", "Cloud Engineer"],
    "Operations": ["Operations Executive", "Operations Manager"],
    "Customer Support": ["Support Engineer", "Customer Success Manager"]
}

countries = {
    "USA": ["Seattle", "New York", "Chicago"],
    "India": ["Hyderabad", "Bangalore", "Chennai", "Mumbai"],
    "UK": ["London", "Manchester"],
    "Canada": ["Toronto", "Vancouver"]
}

employees = []

NUM_EMPLOYEES = 500

for emp_id in range(1, NUM_EMPLOYEES + 1):

    country = random.choice(list(countries.keys()))
    city = random.choice(countries[country])

    department = random.choice(departments)

    title = random.choice(titles[department])

    birth_date = fake.date_of_birth(minimum_age=22, maximum_age=60)

    hire_date = fake.date_between(start_date="-15y", end_date="today")

    salary_ranges = {
        "Sales": (40000, 90000),
        "Finance": (50000, 120000),
        "Human Resources": (45000, 100000),
        "Marketing": (45000, 95000),
        "IT": (70000, 180000),
        "Operations": (45000, 110000),
        "Customer Support": (35000, 70000)
    }

    salary = random.randint(*salary_ranges[department])

    performance = round(random.uniform(2.5, 5.0), 2)

    reports_to = None if emp_id <= 10 else random.randint(1, 10)

    employees.append({

        "employee_id": emp_id,

        "first_name": fake.first_name(),

        "last_name": fake.last_name(),

        "title": title,

        "department": department,

        "birth_date": birth_date,

        "hire_date": hire_date,

        "address": fake.street_address(),

        "city": city,

        "country": country,

        "postal_code": fake.postcode(),

        "phone": fake.phone_number(),

        "email": fake.email(),

        "salary": salary,

        "performance_rating": performance,

        "reports_to": reports_to

    })

df = pd.DataFrame(employees)

df.to_csv("employees.csv", index=False)

print(df.head())

print("\nRows Generated:", len(df))
```

```python

random.seed(42)

order_details = []

NUM_ORDERS = 10000
NUM_PRODUCTS = 1000

for order_id in range(1, NUM_ORDERS + 1):

    # Each order has between 1 and 5 products
    num_items = random.randint(1, 5)

    # Prevent duplicate products within the same order
    products = random.sample(range(1, NUM_PRODUCTS + 1), num_items)

    for product_id in products:

        unit_price = round(random.uniform(10, 500), 2)

        quantity = random.randint(1, 10)

        discount = round(random.choice([0, 0.05, 0.10, 0.15, 0.20]), 2)

        order_details.append({
            "order_id": order_id,
            "product_id": product_id,
            "unit_price": unit_price,
            "quantity": quantity,
            "discount": discount
        })

df = pd.DataFrame(order_details)

df.to_csv("order_details.csv", index=False)

print(df.head())
print(f"\nTotal Rows: {len(df)}")
```
